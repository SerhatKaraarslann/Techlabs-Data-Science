## Code-Snippets

```python
# This is the convention used to import Pandas.
import pandas as pd
```

```python
people = {
    "first": ["Corey", 'Jane', 'John'], 
    "last": ["Schafer", 'Doe', 'Doe'], 
    "email": ["CoreyMSchafer@gmail.com", 'JaneDoe@email.com', 'JohnDoe@email.com']
}
```

```python
# START YOUR CODE HERE.
df = pd.DataFrame(people)
df
```

```python
df.columns # to see the columns in this dataframe
```

```python
df.columns = ['first_name','last_name','email'] # i changed the columns name like that
print(df.columns)
df
```

```python
df.columns = [x.upper() for x in df.columns] # to change column names with list comprehension
df
```

```python
df.columns = df.columns.str.replace(' ','_') # i wanna replace the spaces in column names with underscore, so i use str.replace method
df # nothing change, because my column names have no spaces in their names
#if i do that
#df.columns = df.columns.str.replace('_',' ') # all underscores has been changed with spaces
#df

df.columns = [x.lower() for x in df.columns] # again lower case
df
```

```python
df.rename(columns={'first_name':'first','last_name':'last'}, inplace=True) # if i want to change just some column names, i use the rename method and to save in same dataframe -> inplace = True
df
```

```python
df.loc[2] = ['John','Smith','JohnSmith@email.com'] # I can change an information in a row like that
df

df.loc[2,['last','email']] = ['Doe','JohnDoe@email.com'] # if i want to change more than a column information but not all, i can use this technic
df

df.loc[2,'last'] = 'Smith' # if i wanna change just a value, i dont need a list
df

df.at[2,'last'] = 'Doe' # at method works to change value too
df
```

```python
filt = (df['email'] == 'JohnDoe@email.com')
# df[filt]['last'] = 'Smith' # so i cant change the last name of the person who fit the filter. There are a lot panda operations on background which cause an error 
df.loc[filt,'last'] = 'Smith' # however that works
df
```

```python
df['email'].str.lower() # i changed all columns values into lower case but the values in df hasn't been changed, to do this
df['email'] = df['email'].str.lower() # i have to write it like this way
df
```

```python
# apply, map, applymap,replace methods

#apply method works either on dataframe or series object and apply it every values
df['email'].apply(len) # shows me the every emails length

def update_email(email): # a basic function to update emails into uppercase
    return email.upper()

df['email'].apply(update_email) # we dont need the pass the parameter in update_email function, aplly take this function and use for all email column values, but doesnt change the values in df 
df['email'] = df['email'].apply(update_email) # change the values in df
df

df['email'] = df['email'].apply(lambda x:x.lower()) # instead of a function i can use lambda function in apply method also
df

# apply in df 
df['email'].apply(len) # that how this method works with series(a vector in df)

df.apply(len) # that works with columns, show how many values have the each columns 

# len(df['email']) its the same what apply method does but with every column

df.apply(len, axis='columns') # show the lenght of every columns, how many values every values, horizontal,df.apply(len) vertical and default value for axis is rows

df.apply(pd.Series.min) # shows the minimum values every rows/series

df.apply(lambda x:x.min()) # its the same, lambda works in df with series
```

```python
# applymap only works with dataframe not with series
# apply take the series from dataframe and if i use it like this df.apply(len) but applymap works with every values in df
df.applymap(len) # shows a df and show every values length

df.applymap(str.lower) # makes all values in df lowercase
```

```python
#map method works with series(columns only), it changes the values of series. I can use dict or a function for that
df['first'].map({'Corey':'Chris','Jane':'Mary'}) # the values that i didnt change has been shown up as NaN

# if a use replace method those values arent replaced with NaN(John stays John)
df['first'].replace({'Corey':'Chris','Jane':'Mary'})
```

The answer options were: "Never," "Less than once per year," "Less than once a month but more than once per year," and "Once a month or more often." Complete the *adjust_OpenSourcer*-function so that it can be applied to the *OpenSourcer* column. The goal is for the new *OpenSourcer* column to contain False (boolean) if a respondent answered "Never". Otherwise, it should contain True (boolean).

```python
def adjust_OpenSourcer(answer):
    
    # START YOUR CODE HERE.
    if answer == 'Never':
        return False
    else:
        return True
    # END YOUR CODE HERE.
    
df["OpenSourcer"] = df["OpenSourcer"].apply(adjust_OpenSourcer)
```

#### 6. What is the main difference between *replace* and *map* as Series/DataFrame functions?

with map the values that i don't change is NaN but with replace is the unchanged value.

<details>    
<summary>
    <font size="3" color="darkgreen"><b>Solution (click to expand)</b></font>
</summary>
<p>
    Map assigns NaN to each element that is not present in its mapper dictionary. Replace substitutes only the values that are present in its mapper dictionary, values that are not present remain unchanged. 
</p>
