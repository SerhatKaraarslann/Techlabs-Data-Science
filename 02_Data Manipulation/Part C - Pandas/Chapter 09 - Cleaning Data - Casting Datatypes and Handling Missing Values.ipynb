## Code-Snippets

```python
# This is the convention used to import Pandas.
import pandas as pd
# OTHER LIBRARIES.
import numpy as np
```

```python
people = {
    'first': ['Corey', 'Jane', 'John', 'Chris', np.nan, None, 'NA'], 
    'last': ['Schafer', 'Doe', 'Doe', 'Schafer', np.nan, np.nan, 'Missing'], 
    'email': ['CoreyMSchafer@gmail.com', 'JaneDoe@email.com', 'JohnDoe@email.com', 
              None, np.nan, 'Anonymous@email.com', 'NA'],
    'age': ['33', '55', '63', '36', None, None, 'Missing']
}
```

```python
# START YOUR CODE HERE.
df = pd.DataFrame(people)
df
```

```python
# this function will remove all rows with NaN values
df.dropna()
```

```python
"""
dropna(...): This is the method to remove "Not a Number" (NaN) or null values.
axis='index': This determines the direction.
'index' (or 0) means: Look at the rows.
(The opposite would be axis='columns', which would delete vertical columns).
how='any': This determines the strictness.
'any' means: Delete the row if even a single value is missing.
(The opposite would be how='all', which only deletes the row if absolutely all values are missing).
"""
```

```python
# the dropna() function has parameters to customize its behavior
# for example, you can specify the axis along which to drop the missing values and how to determine if a row or column should be dropped
df.dropna(axis='index',how='any') # axis=0 is index, axis=1 is columns 
```

```python
df.dropna(axis = 'index',how='any',subset=['email']) # only consider the 'email' column when deciding to drop rows
# this will drop rows where 'email' is NaN
```

```python
# we can also use multiple columns in the subset parameter
df.dropna(axis = 'index',how='any',subset=['email','last']) # only consider the 'email' and 'last' columns when deciding to drop rows
# this will drop rows where either 'email' or 'last' is NaN
```

```python
df.replace('NA',np.nan,inplace=True) # replace string 'NA' with actual NaN value
df.replace('Missing',np.nan,inplace=True) # replace string 'Missing' with actual NaN value
df
```

```python
df.dropna() # now this will drop rows with 'NA' or 'Missing' as well since they have been replaced with NaN
# and remove rows with NaN values
```

```python
df.isna() # check which values are NaN
```

```python
df.fillna('MISSING') # fill NaN values with the string 'MISSING'
# you can also fill NaN values with other values like 0 or a specific number
# it is useful when you want to replace missing data with a default value
# especially for numeric columns where calculations are involved
```

```python
df.dtypes # check datatypes of each column
# here all columns are of object datatype
```

```python
# age column should be of numeric datatype
# df['age'].mean() # this will give an error since age column is of object datatype
type(np.nan) # check the type of np.nan and it is float
df['age'] = df['age'].astype('float') # convert age column to float datatype
#df['age'] = df['age'].astype(int) # convert age column to int datatype, this will give an error since there are NaN values in the age column
# because NaN cannot be converted to int directly
```

```python
df.dtypes # check datatypes of each column now age column is of float datatype
```

```python
# now we can calculate the mean of the age column
df['age'].mean() # now this will work and give the mean of age column ignoring NaN values
```

```python
df['YearsCode'].head(10) # check the first 10 values in the 'YearsCode' column
```

```python
#df['YearsCode'].mean() # this will give an error since 'YearsCode' column is of object datatype
df['YearsCode'] = df['YearsCode'].astype('float') # convert 'YearsCode' column to float datatype
# but still error (could not convert string to float : 'Less than 1 year')
```

```python
df['YearsCode'].unique() # check unique values in 'YearsCode' column to see what other non-numeric values are present
# we can see that there is a string value 'Less than 1 year' and 'More than 50 years' which cannot be converted to float
# we need to handle this value before converting the column to float datatype
```

```python
df['YearsCode'].replace({'Less than 1 year': '0', 'More than 50 years': '51'}, inplace=True) # replace these string values with numeric strings
```

```python
# now we can convert the 'YearsCode' column to float datatype
df['YearsCode'] = df['YearsCode'].astype('float') # convert 'YearsCode' column to float datatype
df['YearsCode'].mean() # now this will work and give the mean of 'YearsCode' column ignoring NaN values
```
