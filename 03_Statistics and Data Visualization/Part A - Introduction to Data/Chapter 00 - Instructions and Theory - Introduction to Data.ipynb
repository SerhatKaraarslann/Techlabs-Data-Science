## What is Statistics?
[Link to the video](https://www.coursera.org/learn/understanding-visualization-data/lecture/KeSCz/what-is-statistics)

- Statistics is the science of **collecting, analyzing, presenting, and interpreting data**.
- Statistics helps us to **make decisions in the face of uncertainty**.
- Statistics is used in **various fields** such as business, healthcare, social sciences, and government.
- Everyday examples of statistics include **polls, surveys, and experiments**.
- Statistics is sort of a big tent discipline. It's incorporating new ideas from theory, practice and allied fields all the time.

- Data summarization and visualization are key components of statistics. Data can be overwhelming, and then there is the need to summarize and visualize it effectively to extract meaningful insights. One of the main goals of reduction of data is to make the dataset comprehensible to the human observer. Statistics provides tools and techniques to achieve this goal. A statistician is well trained at utilizing appropriate methods to summarize and visualize data effectively.
- Understanding data is very important, but of course that leads to the need of being able to act on what we learn from data. There are some fields of statistics that focus on decision-making and inference from data.
- Some of the central tasks in statistics is forecasting and prediction. For example, predicting the weather, stock prices, or election outcomes.
- In the data some variables are easy to measure, while others are more difficult. For example, measuring height is easy, but measuring happiness is more difficult. Statistics provides methods to measure and analyze such complex variables.
- Statistics is also concerned with the design of data collection methods. This includes designing surveys, experiments, and observational studies to ensure that the data collected is representative and unbiased.
- Overall, statistics is a powerful tool for making sense of data and making informed decisions based on that data.

- Mathematics is the language of statistics. It provides the foundation for statistical concepts and methods. Computer programming is also essential for statistics, as it allows us to analyze and visualize large datasets efficiently. Machine learning is a subfield of statistics that focuses on developing algorithms that can learn from data and make predictions or decisions based on that data. probability theory is a branch of mathematics that deals with the study of random events and their likelihood of occurrence. It is a fundamental concept in statistics, as it provides the basis for making inferences about populations based on samples. Overall, these fields are interconnected and play a crucial role in the practice of statistics.
- Statistics relies on several other fields, including mathematics, computer programming, machine learning, and probability theory.
